```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../data/raw/creditcard.csv')
print(f'Initial shape: {df.shape}')
```

```python
X = df.drop(['Class', 'Time'], axis=1)
y = df['Class']
```

```python
X_train, X_prov, y_train, y_prov = train_test_split(X,
                                                    y,
                                                    test_size=0.2,
                                                    stratify=y,
                                                    random_state=42)

X_test, X_val, y_test, y_val = train_test_split(X_prov, 
                                                y_prov, 
                                                test_size=0.5, 
                                                random_state=42)
```

```python
X_train_normal = X_train[y_train==0]

X_val_normal = X_val[y_val==0]
```

```python
scaler = StandardScaler()

# Datos para el AE
scaler.fit(X_train_normal)
X_train_normal_scaled = scaler.transform(X_train_normal)
X_val_normal_scaled = scaler.transform(X_val_normal)
X_test_scaled_AE = scaler.transform(X_test)  
X_train_scaled_AE = scaler.transform(X_train)
X_val_scaled_AE = scaler.transform(X_val)
```

```python
np.save('../data/processed/X_test_AE.npy', X_test_scaled_AE)
np.save('../data/processed/X_train_normal.npy', X_train_normal_scaled)
np.save('../data/processed/X_val_normal.npy', X_val_normal_scaled)
np.save('../data/processed/X_train_AE.npy', X_train_scaled_AE)
np.save('../data/processed/X_val_AE.npy', X_val_scaled_AE)
```

```python
# Datos para el Xgboost
scaler2 = StandardScaler()

scaler2.fit(X)

X_train_scaler = scaler2.transform(X_train)
X_val_scaler = scaler2.transform(X_val)
X_test_scaler_XGB = scaler2.transform(X_test)
```

```python
np.save('../data/processed/X_train.npy', X_train)
np.save('../data/processed/X_val.npy', X_val)
np.save('../data/processed/X_test_XGB.npy', X_test_scaler_XGB)
np.save('../data/processed/y_train.npy', y_train)
np.save('../data/processed/y_val.npy', y_val)
np.save('../data/processed/y_test.npy', y_test)
```
