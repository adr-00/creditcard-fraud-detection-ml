```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
DATA = '../data/raw/creditcard.csv'
df = pd.read_csv(DATA)
df
```

```python
plt.figure(figsize=(5,4))
sns.countplot(data=df, x='Class')
plt.title('Classes distribution')
plt.show()
```

```python
fraud_ratio = df["Class"].mean() * 100
print(f"\n% de fraudes: {fraud_ratio:.4f}%")
```

```python
fig, ax = plt.subplots(1,2,figsize=(12,4))

sns.histplot(df['Amount'], bins=50, ax=ax[0])
ax[0].set_title('Distribution: Amount')

sns.histplot(df['Time'], bins=50, ax=ax[1])
ax[1].set_title('Distribution: Time')

plt.show()
```

Correlation between each feature and 'class', we can see that the correlation betwenn each feature is near to 0, so we cant garantee that there are linear correlation

```python
corr_class = df.corr(numeric_only=True)['Class'].sort_values(ascending=False)
corr_class
```

```python
top_corr = df.corr(numeric_only=True)['Class'].abs().sort_values(ascending=False).head(10).index
top_corr
```

```python
sns.heatmap(df[top_corr].corr(), annot=False, cmap='coolwarm')
plt.title('Most correlated features with Class')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,4))
sns.boxplot(x=df["Amount"], ax=ax[0])
ax[0].set_title("Boxplot Amount")

sns.boxplot(x=df["Time"], ax=ax[1])
ax[1].set_title("Boxplot Time")

plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,4))

sns.kdeplot(data=df, x="Amount", hue="Class", ax=ax[0], fill=True)
ax[0].set_title("Distribución Amount según clase")

sns.kdeplot(data=df, x="Time", hue="Class", ax=ax[1], fill=True)
ax[1].set_title("Distribución Time según clase")

plt.show()
```
